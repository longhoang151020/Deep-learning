### Nhận xét & Giải thích - Bài 6
* **Phân biệt `a * b` với `a @ b.T`:**
  * `a * b`: Là phép nhân **Element-wise** (nhân từng phần tử tương ứng). Cả 2 tensor phải cùng shape $(2, 3)$, kết quả ra shape $(2, 3)$.
  * `a @ b.T`: Là phép **nhân ma trận đại số tuyến tính**. Ma trận $a$ kích thước $(2, 3)$ nhân với ma trận chuyển vị $b^T$ kích thước $(3, 2)$.
* **Shape của `a @ b.T`:** `torch.Size([2, 2])`.

````python
import torch

a = torch.randn(3, 4)
print("Tensor a:\n", a)

print("\n--- Khi dim=0 (Tính theo cột) ---")
print("Mean:", torch.mean(a, dim=0))
print("Std:", torch.std(a, dim=0))

print("\n--- Khi dim=1 (Tính theo hàng) ---")
print("Mean:", torch.mean(a, dim=1))
print("Std:", torch.std(a, dim=1))
print("Cumulative sum:\n", torch.cumsum(a, dim=1))
````

### Nhận xét & Giải thích - Bài 7
* **Nhận xét sự khác nhau giữa `dim=0` và `dim=1`:**
  * `dim=0`: Thực hiện tính toán **theo từng cột** (tính dọc theo các hàng). Trả về tensor có 4 phần tử (tương ứng 4 cột).
  * `dim=1`: Thực hiện tính toán **theo từng hàng** (tính ngang qua các cột). Trả về tensor có 3 phần tử (tương ứng 3 hàng).

````python
import torch

x1 = torch.tensor(3.6, requires_grad=True)
y1 = x1 * x1
y1.backward()
print("Hàm y = x^2:")
print("x.grad (PyTorch):", x1.grad)

x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2**2 + 2 * x2 + 1
y2.backward()
print("\nHàm y = 3*x^2 + 2*x + 1:")
print("x.grad (PyTorch):", x2.grad)
````

### Nhận xét & Giải thích - Bài 8
* **Tính thủ công đạo hàm $y = x^2$ tại $x = 3.6$:**
  * Đạo hàm lý thuyết: $y' = \frac{dy}{dx} = 2x$.
  * Tại $x = 3.6 \Rightarrow y' = 2 \times 3.6 = 7.2$.
  * **So sánh:** Kết quả tính thủ công ($7.2$) trùng khớp hoàn toàn với `x.grad` từ PyTorch ($7.2000$).
* **Kiểm tra hàm mới $y = 3x^2 + 2x + 1$ tại $x = 3.6$:**
  * Đạo hàm lý thuyết: $y' = 6x + 2$.
  * Tại $x = 3.6 \Rightarrow y' = 6(3.6) + 2 = 21.6 + 2 = 23.6$. Kết quả PyTorch trả về đúng $23.6$.

````python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1)
Y = polynomial(X)

XX = np.hstack((X * X, X, np.ones_like(X)))
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
w = torch.randn(3, 1, requires_grad=True)

optimizer = torch.optim.NAdam([w], lr=0.01)

for epoch in range(1000):
    y_pred = x @ w
    mse = torch.mean(torch.square(y_pred - y))
    optimizer.zero_grad()
    mse.backward()
    optimizer.step()

print("Learned coefficients (Hệ số học được):\n", w.detach().numpy().flatten())
print("Target coefficients (Hệ số mục tiêu): [1. 2. 3.]")
````

### Nhận xét & Giải thích - Bài 9
* **So sánh 3 hệ số học được:** Trọng số `w` ban đầu khởi tạo ngẫu nhiên, sau 1000 epoch huấn luyện đã hội tụ về rất sát với hệ số mục tiêu thực tế `[1, 2, 3]`.
* **Giải thích vai trò của các hàm:**
  * `optimizer.zero_grad()`: Xóa sạch các giá trị gradient đã lưu ở bước trước. Nếu không reset, gradient sẽ bị cộng dồn khiến quá trình cập nhật bị sai lệch.
  * `mse.backward()`: Thực hiện lan truyền ngược (backpropagation) để tự động tính gradient của hàm tổn thất (MSE) theo các trọng số `w` và lưu vào `w.grad`.
  * `optimizer.step()`: Cập nhật trọng số `w` dựa trên giá trị gradient vừa tính được theo thuật toán tối ưu NAdam.

````python
import random
import torch

EPOCHS = 2000
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print(f"A = {A.item():.4f}, B = {B.item():.4f}, C = {C.item():.4f}, D = {D.item():.4f}")
print("Kiểm tra A + B (Mục tiêu 9):", (A + B).item())
print("Kiểm tra C - D (Mục tiêu 1):", (C - D).item())
print("Kiểm tra A + C (Mục tiêu 8):", (A + C).item())
print("Kiểm tra B - D (Mục tiêu 2):", (B - D).item())
````

### Nhận xét & Giải thích - Bài 10
* **Kiểm tra nghiệm:** Thay bộ nghiệm $A, B, C, D$ tìm được vào 4 phương trình ban đầu cho kết quả sai số xấp xỉ 0.
* **Nhận xét vì sao bài toán có nhiều nghiệm:**
  * Ta có 4 phương trình: (1) $A+B=9 \Rightarrow B=9-A$, (2) $C-D=1 \Rightarrow D=C-1$, (3) $A+C=8 \Rightarrow C=8-A$, (4) $B-D=2$.
  * Thế các ẩn theo $A$ vào (4): $(9-A) - (8-A-1) = 2 \Leftrightarrow 2 = 2$ (luôn đúng với mọi $A$).
  * Phương trình (4) phụ thuộc tuyến tính vào 3 phương trình còn lại, nên hệ chỉ có 3 phương trình độc lập cho 4 ẩn $\Rightarrow$ Tồn tại **vô số nghiệm**. Mô hình sẽ hội tụ về nghiệm nào tùy thuộc vào giá trị khởi tạo ban đầu.

````python
import torch
import numpy as np

print("=== Câu 1: Reshape Tensor X (24 phần tử) ===")
X = torch.arange(24)
print("Reshape 4x6:\n", X.reshape(4, 6))
print("Reshape 2x12:\n", X.reshape(2, 12))
print("Reshape 2x3x4:\n", X.reshape(2, 3, 4))

print("\n=== Câu 2: Các phép toán trên A, B (3x4) ===")
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("A + B:\n", A + B)
print("A * B:\n", A * B)
print("A / B:\n", A / B)
print("Trung bình A theo cột:", torch.mean(A, dim=0))
print("Độ lệch chuẩn A theo cột:", torch.std(A, dim=0))

print("\n=== Câu 3: Autograd cho y = 2x^2 + 5x + 3 tại x = 2 ===")
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()
print("dy/dx tại x=2 (PyTorch Autograd):", x.grad.item())
print("dy/dx tính thủ công (4*2 + 5):", 4 * 2 + 5)

print("\n=== Câu 4: Tìm hệ số cho y = 2x^2 - 3x + 5 ===")
poly = np.poly1d([2, -3, 5])
N = 30
X_np = np.random.randn(N, 1)
Y_np = poly(X_np)

XX = np.hstack((X_np * X_np, X_np, np.ones_like(X_np)))
x_ten = torch.tensor(XX, dtype=torch.float32)
y_ten = torch.tensor(Y_np, dtype=torch.float32)
w = torch.randn(3, 1, requires_grad=True)

opt = torch.optim.NAdam([w], lr=0.01)
for epoch in range(1500):
    pred = x_ten @ w
    loss = torch.mean(torch.square(pred - y_ten))
    opt.zero_grad()
    loss.backward()
    opt.step()

print("Hệ số học được [a, b, c]:", w.detach().numpy().flatten())
print("Hệ số thực tế: [ 2. -3.  5.]")
````
